```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score,root_mean_squared_error
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv("../data/raw/Salary_Data.csv")
```

```python
df.head()
```

```python
x=df[['YearsExperience']]
y=df[['Salary']]
```

```python
plt.figure(figsize=(3,3))
sns.pairplot(df, x_vars=['YearsExperience'], y_vars=['Salary'], size = 3, kind="scatter")
plt.title("salary vs years")
plt.xlabel("years")
plt.ylabel("salary")
plt.show()
 
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_train)
```

```python
reg_line=LinearRegression()
reg_line.fit(x_train,y_train)
## coef_ : weight 
print(reg_line.coef_)
print(reg_line.intercept_)
y_hat_train=reg_line.predict(x_train)
y_hat_test=reg_line.predict(x_test)
```

```python
print(y_hat_train[:10])
print(y_train[:10])
```

```python
plt.style.use('dark_background')
plt.figure(figsize=(9,6))
plt.scatter(x_train,y_train,c='b')
plt.plot(x_train,y_hat_train,c='r')
plt.title("Salary vs Experience (Training Set)")
plt.xlabel("Experience ")
plt.ylabel("Salary")
```

```python
plt.style.use('dark_background')
plt.figure(figsize=(9,6))
plt.scatter(x_test,y_test,c='b')
plt.plot(x_test,y_hat_test,c='r')
plt.title("Salary vs Experience (Testing Set)")
plt.xlabel("Experience ")
plt.ylabel("Salary")
```

```python
# calculate mean squred error
mse_test=mean_squared_error(y_test,y_hat_test)
mse_train=mean_squared_error(y_train,y_hat_train)
print(mse_train)
print(mse_test)
rmse=root_mean_squared_error(y_test,y_hat_test)
r2_score=r2_score(y_test,y_hat_test)
print(rmse)
```
