```python
class LinearRegressionGD:
    def __init__(self,x:np.array,y:np.array,learning_rate :float =0.01,n_iters : int=100):
        self.x=x
        self.y=y
        self.learning_rate=learning_rate
        self.n_iters=n_iters
        self.w=0
        self.b=0
        self.n=len(self.x)
        self.SSE=[]
        

    def fit(self):
        for i in range(self.n_iters):
            y_hat= self.w  *  self.x  + self.b
            error=y_hat-self.y

            # mse= np.sum((error)**2)/self.n
            self.SSE.append(np.sum((error)**2))

            dw= np.sum((error)*self.x)/self.n
            db= np.sum((error))/self.n

            self.w=self.w-self.learning_rate*dw
            self.b=self.b-self.learning_rate*db

    def get_theta_1(self):
        return self.w

    def get_theta_0(self):
        return self.b


    def predict(self,x):
        return self.w*x+self.b

    def plot_training(self):

        plt.figure(figsize=(9,6))
        plt.subplot(1,2,1)
        plt.plot(range(1,self.n_iters+1),self.SSE)
        plt.xlabel("Iterations")
        plt.ylabel("SSE")
        plt.title("SSE over Iterations")
        plt.scatter(self.x, self.y) 
        plt.subplot(1,2,2)
        plt.plot(self.x,self.predict(self.x))
        
        plt.xlabel("X")
        plt.ylabel("Y")
        plt.title("Regression Line")

    def calc_MSE(self,y,y_predicted):
        return np.sum((y_predicted-y)**2)/len(y)



    


    
```
