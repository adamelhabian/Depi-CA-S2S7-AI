## my task is write first two iteration in linear regression algorithm using gradient descent and MSE then compare my manual answer with result come from code 

```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
x=np.array([1,2,3,4])
y=np.array([2,2.8,3.6,4.5])
```

```python
w=0
b=0
alpha=0.01
num_iteration=2
error_values=[]
```

```python
for i  in range(num_iteration):

    y_hat=x*w+b

    error= (y_hat-y)
    cost= np.sum((error)**2)/len(x)
    error_values.append(cost)

    w_der=  (2* np.sum((error)*x))/len(x)
    b_der= (2* np.sum((error)))/len(x)

    w = w-alpha*w_der
    b=b-alpha*b_der
    print("w : ",w)
    print("b : ",b)

print(error_values)    
```
