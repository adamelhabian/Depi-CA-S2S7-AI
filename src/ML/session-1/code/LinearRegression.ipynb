```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
x=np.array([1,2,3,4,5])
y=np.array([2,2.2,2.8,3.6,4])
```

```python
w=0
b=0
alpha=0.01
num_iteration=100
error_values=[]
print(len(x))
```

```python
for i in range(num_iteration):
    y_hat=w*x+b

    D_w= ( 2 * np.sum((y_hat-y)*x))/  (2* len(x))
    D_b= 2* np.sum(y_hat-y) / 2*len(x)

    w=w-alpha*D_w
    b=b-alpha*D_b

    cost = (np.sum((y_hat-y)**2))/len(x)
    error_values.append(cost)



print(error_values)    
```

```python
plt.figure(figsize=(9,5))

plt.subplot(1,2,1)
plt.plot(range(num_iteration),error_values)
```
