```python
def automated_stat_analyzer(df, column_name):
    """
    Company Task: Provide a summary report of a specific data variable.
    
    Instructions:
    1. Check if the column is numerical or categorical.
    2. For numerical: Calculate Mean, Median, and Standard Deviation.
    3. For categorical: Calculate the Mode.
    4. Return a dictionary with these statistical measures.
    """
    # Implement using df[column_name].mean(), .median(), .std(), or .mode()  you can used Sales_Amount for your test case
    if(pd.api.types.is_numeric_dtype(df[column_name])):
        result={
        "mean" :df[column_name].mean(),
        "median": df[column_name].median(),
        "std"  : df[column_name].std(),
        "Is_Skewed": df[column_name].mean() != df[column_name].median()
        
        }
    elif pd.api.types.is_categorical_dtype(df[column_name]):
        result={
            "mode" :df[column_name].mode()
        }
    return result

```

```python
result=automated_stat_analyzer(df_test,"Sales_Amount")
print(result)
print(result['mean'])
print(result['median'])
print(result['std'])
print(result['Is_Skewed'])

```

```python
def null_handling_strategy(df, strategy="fill_mean"):
    """
    Company Task: Clean a dataset by resolving missing (NaN) values.
    """
    if(df.isnull().values.any() == False): return
    elif strategy=='drop_rows': return df.dropna(inplace=True)
    else:
        null_columns=list(df.columns[df.isnull().any()])
        for  i in range (len(null_columns)):
            if pd.api.types.is_numeric_dtype(df[null_columns[i]]):
                if (strategy=='fill_mean'):
                    df[null_columns[i]]= df[null_columns[i]].fillna(df[null_columns[i]].mean())
                    return df[null_columns[i]]
                elif (strategy=='fill_median'):
                    df[null_columns[i]]= df[null_columns[i]].fillna(df[null_columns[i]].median())
                    return df[null_columns[i]]

                
```

```python
print(null_handling_strategy(df_test))
```
