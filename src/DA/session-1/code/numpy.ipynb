````python
def array_reshape(arr : list,rows : int,cols :int) -> list:
    try:
        arr=np.array(arr)
        return arr.reshape(rows,cols)
    except(TypeError,ValueError) as e:
        raise ValueError(f"Invalid reshape operation: {e}")
````

````python
vec=[1,2,3,4,5,6]

array_reshape(vec,2,3)
````

### Saving and Loading Multiple NumPy Arrays


````python
mat1=np.ones((5,3))
mat2=np.zeros((5,3))
np.savez("np_file.npz",cat=mat1,dog=mat2)
````

````python
with np.load("np_file.npz") as file:
    mat1=file['cat']
    mat2=file['dog']

print(mat1)
print(mat2)    
````

### `np.isclose()`

`np.isclose()` checks whether two values or arrays are **close enough to each other** within a specified tolerance.

```python
np.isclose(v1, v2, atol=3)

````python
v1=np.array([1,2,3,4,5])
v2=np.array([2,4,3,4,10])
np.isclose(v1,v2,atol=3)

# print(np.dot(v1,v2)/(np.linalg.det(v1)*np.linalg(v2)))

````
