### 3. Write a program to find out the prime factors of a number. Example: prime factors of 56 - 2, 2, 2, 7

```python

def is_prime(num)-> bool:
    '''
    this function check number is prime or not
    arg:
    num: number we want to check
    '''
    for i in range(2,int(num**0.5)+1):
        if(num % i == 0):
            return False    
    return True        


def primeFactors(num):
    
    out=[]
    if(is_prime(num)):
        out.append(num)
        return out
    while(is_prime(num)==False):

        if(num % 2 ==0):
            num//=2
            out.append(2)
        else:
            if(num %3==0):
                num//=3
                out.append(3) 
            # Composite number like 25,35 , 49    
            else:  
                for i in range(3,int(num **0.5)+1):
                    if(num % i == 0):
                        num//=i
                        out.append(i)
                        break

    out.append(num)
    return out               


print(primeFactors(56))
```

### 4. Write a function that converts a decimal number to binary number

```python
def decToBin(num):

    result=""
    if(num==0):
        return "0"
    
    while(num !=0):
        
        if(num%2 !=0):
            result="1"+result
        else:
            result="0"+result
        num=num//2
    return result


print(decToBin(11))

```
