# Functions

- **Function naming:** Use `lowercase` / `snake_case` for functions. Use `CamelCase` for classes.
- **Docstrings:** Use docstrings to document what the function does, its parameters, and return value.
- **Type hints:** Type hints describe the expected data types but do not affect or enforce the code at runtime.
- **Return type:** `-> str` means the function is expected to return a string, but Python does not force the return type.
- **Default values:** You can assign a default value to a parameter using `=`. Example: `name: str = "Unknown"`.
- **Parameter order:** Required (non-default) parameters must come before parameters with default values.
- **Optional type hint:** Use `Optional[type]` when a parameter can have a value of that type or `None`. Example: `name: Optional[str] = None`.

```python
# Hyber parameters : parameters that i adjust on it 
# None is a value not datatype


#                var then type hint then   default value
def info_user( name : str|None="adam" ,age : int = 20) ->str:
    '''
    this function will return the name and age of the user   
    Args:
        name (str) :name of the user
        age (int) : age of the user
    '''
    return f"my name is {name} and my age is {age}"

 
print(info_user(age=33))
print(info_user())
```

# Varadic Argument Function
- **Variadic arguments:** Allow a function to accept a variable number of arguments using `*args` for positional arguments and `**kwargs` for keyword arguments.

```python
# * -> make tuple , ** -> make dictionary
def func_1(*name):
    print(name)
    print(type(name))

func_1("adam","ahmed","elhabian",123)    


# GENATIC ALGORITHM  
```

```python
def func_2(**name):
    print(name)
    print(type(name))

func_2(name="adam",age=20)
```

# Real Example 

```python
def train_model(model_name,*datasets , **hyperparameters):
    print(f"Training Model : {model_name}")


    print("Datasets :  ")
    for dataset in datasets:
        print("-",dataset )

    print("\nHyperparameters")

    for key,value in hyperparameters.items():    
        print(f"{key} : {value}")


train_model("Fraud Detection" ,"train.csv","Validation.csv","test.csv",
            learning_rate=0.01,
            epochs=100,
            batch_size=64,
            optimizer="Adam",
            early_stopping=True)
    
```

# Lambda Function
- **Lambda function:** A small anonymous function used for simple, one-line operations. Syntax: `lambda arguments: expression`.

```python
def sum(x):
    return x + 10 

sum(5)
```

```python
sum_l=lambda x : x+10
print(sum_l(5))
type(sum_l)
```

```python
def function_1(n):
    return lambda a : 2*n

fucntion_2=function_1(2)
```

```python
transformers= {
    "lower": lambda text:text.lower(),
    "strip":lambda text:text.strip()
}

text="            Machine Learning With Python  "

print(transformers["strip"](text))
print(text)

print(transformers['lower'](text))
```
