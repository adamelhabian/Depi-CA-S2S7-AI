## Task: Remove `"python"` from a List

### Steps:

1. Read the size of the list from the user.
2. Read the list items from the user.
3. Check if `"python"` exists in the list.
4. If it exists, remove its **first occurrence**.
5. Otherwise, print:
   ```
   Input list Doesn't contain python
   ```
6. Print the updated list.

````python

size= int(input("Enter size of list"))

input_list=[]
for i in range(size):
    item=input(f"Enter item {i+1} :")
    input_list.append(item)

if "python" in input_list: input_list.remove("python")
else : print("Input list Doesn't contain python ")
print(input_list)
````
