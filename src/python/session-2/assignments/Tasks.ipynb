# Replace the last value of each tuple in a list

- Sample list:
- [(10, 20, 40), (40, 50, 60), (70, 80, 90)]
- Expected output:
- [(10, 20, 100), (40, 50, 100), (70, 80, 100)]

```python
data = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]

# TODO: replace last value with 100
new_list = []
# your code here
for t in data:
    temp_list = list(t)
    temp_list[-1] = 100
    temp_tuple = tuple(temp_list)
    new_list.append(temp_tuple)

print(new_list)

# Hint: Convert each tuple to list, edit, convert back.
```

# Find the longest word in a list

```python
words = ["hi", "hello", "fantastic", "world"]

# TODO: find the longest word
longest=words[0]
for word in words:
    if(len(word)>len(longest)):longest=word

print(longest)
# Hint: Track the word with the largest len().
```
