
# Python Docstrings

<details>
<summary>1. What is a Docstring?</summary>

A **docstring (documentation string)** is a special string in Python used to document:

- Functions
- Classes
- Modules
- Methods

It explains what the code does and helps developers understand how to use it.

Example:

```python
def greet(name):
    """
    Say hello to a user.

    Args:
        name (str): User name.

    Returns:
        str: Greeting message.
    """
    return f"Hello {name}"
```

</details>


<details>
<summary>2. Why Do We Use Docstrings?</summary>

Docstrings help with:

- Code readability
- Project maintenance
- Team collaboration
- Automatic documentation generation
- Understanding APIs and libraries

They are commonly used in:

- Machine Learning projects
- Data Science projects
- Backend applications
- Open-source packages

</details>


<details>
<summary>3. Docstring Rules</summary>

### Rule 1: Must be the first statement

Correct:

```python
def test():
    """
    Test function.
    """
```

Incorrect:

```python
def test():
    print("Hello")
    """
    Test function.
    """
```

---

### Rule 2: Use triple quotes

Recommended:

```python
"""
Documentation here
"""
```

---

### Rule 3: Explain purpose, not implementation

Good:

```python
"""
Calculate the average score.
"""
```

Avoid:

```python
"""
Loop over values and divide by length.
"""
```

</details>


<details>
<summary>4. Docstring Formats</summary>

## Google Style

```python
def calculate_accuracy(y_true, y_pred):
    """
    Calculate model accuracy.

    Args:
        y_true (list): Actual labels.
        y_pred (list): Predicted labels.

    Returns:
        float: Accuracy score.
    """
```

</details>


<details>
<summary>5. Accessing Docstrings</summary>

Using:

```python
function_name.__doc__
```

Example:

```python
print(calculate_accuracy.__doc__)
```

or:

```python
help(calculate_accuracy)
```

</details>


<details>
<summary>6. Docstrings in Machine Learning Projects</summary>

Example:

```python
def preprocess_data(df):
    """
    Clean dataset before model training.

    Args:
        df (DataFrame): Raw dataset.

    Returns:
        DataFrame: Processed dataset.
    """
```

Useful for documenting:

- Data preprocessing
- Feature engineering
- Model training
- Evaluation functions
- Pipelines

</details>


<details>
<summary>7. Summary</summary>

## Key Points:

- Docstrings document Python code.
- They are written using triple quotes.
- They must be the first statement.
- They describe purpose, inputs, and outputs.
- They improve readability and maintainability.

### Remember:

> Comments explain specific code lines.  
> Docstrings explain how to use your code.

</details>

# Python Function Sources & NameError

<details>
<summary>1. Where Does a Function Come From?</summary>

In Python, a function can come from:

- **Built-in functions** → Provided by Python (`print()`, `len()`)
- **Module functions** → Defined inside a `.py` file
- **Class methods** → Defined inside a class
- **User-defined functions** → Created in your current file

</details>


<details>
<summary>2. NameError</summary>

`NameError` happens when Python cannot find the name you are using.

Example:

```python
calculate()
```

Error:

```text
NameError: name 'calculate' is not defined
```

Meaning:
- Function was not created
- Function was not imported
- Function name has a typo

</details>


<details>
<summary>3. Module Function (.py File)</summary>

A **module** is a Python file (`.py`) that contains functions.

Example:

`math_operations.py`

```python
def add(a, b):
    return a + b
```

Usage:

```python
from math_operations import add
```

The function source is the module:

```python
print(add.__module__)
```

</details>


<details>
<summary>4. Class Method</summary>

A function inside a class is called a **method**.

Example:

```python
class Calculator:
    def add(self, a, b):
        return a + b
```

`add()` belongs to the class:

```
Calculator.add
```

</details>


<details>
<summary>5. How To Check Function Source?</summary>

Check module:

```python
function.__module__
```

Check class:

```python
function.__qualname__
```

Example:

```python
print(len.__module__)
```

Output:

```
builtins
```

</details>


<details>
<summary>6. Quick Comparison</summary>

| Type | Source | Example |
|---|---|---|
| Built-in | Python | `len()` |
| Module Function | `.py` file | `math.sqrt()` |
| Method | Class | `obj.method()` |
| Unknown | Not found | `NameError` |

</details>


<details>
<summary>Summary</summary>

- `.py file` containing functions → **Module**
- Function inside class → **Method**
- Python provided functions → **Built-in**
- Function not found → **NameError**

Remember:

```
Module (.py) → Functions
Class       → Methods
Python      → Built-ins
Unknown     → NameError
```

</details>

# Python Data Structures Comparison: List vs Tuple vs Set vs Dictionary

<details>
<summary>1. Quick Comparison Table</summary>

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Syntax | `[ ]` | `( )` | `{ }` | `{key: value}` |
| Ordered | ✅ Yes | ✅ Yes | ❌ No | ✅ Yes (Python 3.7+) |
| Mutable | ✅ Yes | ❌ No | ✅ Yes | ✅ Yes |
| Allows Duplicates | ✅ Yes | ✅ Yes | ❌ No | Keys ❌ / Values ✅ |
| Indexed | ✅ Yes | ✅ Yes | ❌ No | By key |
| Stores | Values | Values | Unique values | Key-value pairs |
| Fast Lookup | Slow | Slow | Fast | Very Fast |
| Use Case | Collection of items | Fixed data | Unique items | Data mapping |

</details>
