## Encoded Message:

`&&&**$gnirtS PLIO!!@1234`

## Steps to Decode:

1. Extract the core part of the message: `"gnirtS PLIO"`.

2. Reverse the first word: `"gnirtS"` becomes `"String"`.

3. Replace shifted vowels in the second word:
   - `"PLIO"`: Replace `I -> E` and `O -> U` to get `"PLEU"`.

4. Final decoded message: `"String PLEU"`.

```python
message="&&&**$gnirtS PLIO!!@1234"

core=message[6:17:1]
print("Core Message :",core)
core=core.replace("gnirtS","String")
print("After Replace First Word by String :",core)
core = core.replace("I", "E")
core = core.replace("O", "U")

print("Output : ",core)
```
