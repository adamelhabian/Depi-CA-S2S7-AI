## Encoded Message:

`###!!@mocleW EPGTQ!!!6789`

## Steps to Decode:

1. Extract the core part of the message: `"mocleW EPGTQ"`.
2. Reverse the first word: `"mocleW"` becomes `"Welcome"`.
3. Replace shifted vowels in the second word:
   - `"EPGTQ"`: No vowels to change.
4. Final decoded message: `"Welcome PGTQ"`.

```python
Message="###!!@mocleW EPGTQ!!!6789"

# Steps to Decode 

core=Message[6:18:1]
print("Core Part : ",core)
core=core.replace("mocleW","Welcome")
print("Replaced message :",core)
second_word= core[8:]

vowels="aeiou"
result=core[0:8]

for c in second_word:
    if c.lower() not in vowels: result+=c

print("Output : ",result)


```
