### Class 1 

```python
class MultiplicationTable :
    '''
    MultiplicationTable class take number and represents multiplication table of that number 
    
    Attributes:
        number (int) : The number to be checked
    '''

    def __init__(self,number=1):
        self.number=number

    def display_table(self):
        '''
        Display the multiplication table of the number from 1 to 10
        
        Returns:
            None
        '''
        for i in range(10):
            print(f"{self.number} * {i+1} = {self.number*(i+1)}")

input_number=int(input("Enter The Number"))
m= MultiplicationTable(input_number)           
m.display_table()
```

### Class 3

```python
class PrimeFactors:
    ''' 
    Represent The Prime Factors of number

    Attributes:
        num (int) : The number that we find out it's prime factors
    '''
    def __init__(self,num):
        self.num=num

    def is_prime(self,num)-> bool:
        '''
        this function check number is prime or not
        arg:
        num: number we want to check
        '''
        if num <2 : return False
        for i in range(2,int(num**0.5)+1):
            if(num % i == 0):
                return False    
        return True 

    
    def get_primeFactors(this)-> list:
        '''
          return prime factors of number (list)  
        '''
    
        out=[]
        if(this.is_prime(this.num)):
            out.append(this.num)
            return out
        while(this.is_prime(this.num)==False):

            if(this.num % 2 ==0):
                this.num//=2
                out.append(2)
            else:
                if(this.num %3==0):
                    this.num//=3
                    out.append(3) 
                # Composite number like 25,35 , 49    
                else:  
                    for i in range(3,int(this.num **0.5)+1):
                        if(this.num % i == 0):
                            this.num//=i
                            out.append(i)
                            break 

        out.append(this.num)
        return out  
    def display_primeFactors(self):
        print(f"Prime Factors of Number {self.num} is {self.get_primeFactors()} ")    

pf = PrimeFactors(56)
pf.display_primeFactors()


```

### Class 4

```python
class DecimalToBinary:
    '''  
    Covert Decimal number to binary number 

    Attributes:
        number (int) : The Decimal number we want to convert it to Binary number
    '''

    def __init__(self,number):
        self.num = number

    def get_binary(self)->str:
        '''
        return str of Binary Numbers 
        '''
        num=self.num
        result=""
        if(num==0):
            return "0"
        
        while(num !=0):
            
            result = str(num % 2) + result
            num=num//2
        return result
    def display_binary(self):
        '''
            display Bianry number 

            return:
                None
        '''
        print(self.get_binary())

db=DecimalToBinary(11)
db.display_binary()        

        
```

```python
class PerfectNumbers:
    '''
    Reprecent all perfect numbers in a given range .
    A number is called perfect if the sum of proper divisors of that number is equal to the  number
    Attributes :
        start (int): The Start of range
        end (int) : The End of Range

    '''
    def __init__(self,start,end):
        self.start=start
        self.end=end

    def  display_perfect_nums(self):

        '''
        Display the perfect numbers within the range   
        '''
        for num in range(self.start, self.end + 1):
                if num < 2:
                    continue
        
                divisor_sum = 1
        
                for i in range(2, int(num ** 0.5) + 1):
                    if num % i == 0:
                        divisor_sum += i
        
                        if i != num // i:
                            divisor_sum += num // i
        
                if divisor_sum == num:
                    print(num)


pn=PerfectNumbers(1,100)
pn.display_perfect_nums()
            
```
