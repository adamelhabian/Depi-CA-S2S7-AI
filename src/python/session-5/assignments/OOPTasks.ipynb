```python
class MultiplicationTable :
    '''
    MultiplicationTable class take number and represents multiplication table of that number 
    
    Attributes:
        number (int) : The number to be checked
    '''

    def __init__(self,number=1):
        self.number=number

    def display_table(self):
        '''
        Display the multiplication table of the number from 1 to 10
        
        Returns:
            None
        '''
        for i in range(10):
            print(f"{self.number} * {i+1} = {self.number*(i+1)}")

input_number=int(input("Enter The Number"))
m= MultiplicationTable(input_number)           
m.display_table()
```

```python
class PrimeFactors:
    ''' 
    Represent The Prime Factors of number

    Attributes:
        num (int) : The number that we find out it's prime factors
    '''
    def __init__(self,num):
        self.num=num

    def is_prime(self,num)-> bool:
        '''
        this function check number is prime or not
        arg:
        num: number we want to check
        '''
        if num <2 : return False
        for i in range(2,int(num**0.5)+1):
            if(num % i == 0):
                return False    
        return True 

    
    def get_primeFactors(this)-> list:
        '''
          return prime factors of number (list)  
        '''
    
        out=[]
        if(this.is_prime(this.num)):
            out.append(this.num)
            return out
        while(this.is_prime(this.num)==False):

            if(this.num % 2 ==0):
                this.num//=2
                out.append(2)
            else:
                if(this.num %3==0):
                    this.num//=3
                    out.append(3) 
                # Composite number like 25,35 , 49    
                else:  
                    for i in range(3,int(this.num **0.5)+1):
                        if(this.num % i == 0):
                            this.num//=i
                            out.append(i)
                            break 

        out.append(this.num)
        return out  
    def display_primeFactors(self):
        print(f"Prime Factors of Number {self.num} is {self.get_primeFactors()} ")    

pf = PrimeFactors(56)
pf.display_primeFactors()


```

### Class 4

```python
class DecimalToBinary:
    '''  
    Covert Decimal number to binary number 

    Attributes:
        number (int) : The Decimal number we want to convert it to Binary number
    '''

    def __init__(self,number):
        self.num = number

    def get_binary(self)->str:
        '''
        return str of Binary Numbers 
        '''
        num=self.num
        result=""
        if(num==0):
            return "0"
        
        while(num !=0):
            
            result = str(num % 2) + result
            num=num//2
        return result
    def display_binary(self):
        '''
            display Bianry number 

            return:
                None
        '''
        print(self.get_binary())

db=DecimalToBinary(11)
db.display_binary()        

        
```

```python
class PerfectNumbers:
    '''
    Reprecent all perfect numbers in a given range .
    A number is called perfect if the sum of proper divisors of that number is equal to the  number
    Attributes :
        start (int): The Start of range
        end (int) : The End of Range

    '''
    def __init__(self,start,end):
        self.start=start
        self.end=end

    def  display_perfect_nums(self):

        '''
        Display the perfect numbers within the range   
        '''
        for num in range(self.start, self.end + 1):
                if num < 2:
                    continue
        
                divisor_sum = 1
        
                for i in range(2, int(num ** 0.5) + 1):
                    if num % i == 0:
                        divisor_sum += i
        
                        if i != num // i:
                            divisor_sum += num // i
        
                if divisor_sum == num:
                    print(num)


pn=PerfectNumbers(1,100)
pn.display_perfect_nums()
            
```

```python
class Bank_Account:
    '''
    Represent Banking Account with Intial Balance =1000

    Attribute:
        amount (float) : represent the amount of balance in your account
    '''
    def __init__(self,amount=1000):
        self.balance=amount

    def withdraw(self,amount):
        '''
        Withdraw amount of money   
        '''
        if(amount<=self.balance):
            self.balance-=amount
            print(f"Withdraw {amount} . New balance is {self.balance}")
        else:
            print("Insufficient funds")

    def deposit(self,amount):
        '''
        Deposit amount of money 
        '''
        self.balance+=amount
        print(f"Deposited {amount} . New balance is {self.balance}")

    def check_balance(self):
        print("Current balance is :",self.balance)  



b=Bank_Account(1000)
b.withdraw(100)
b.deposit(1000)
b.check_balance()



    
```

Create an abstract class Animal and derive specific animal classes from it. Each derived class should implement the abstract methods defined in the Animal class.

Instructions:

- Create an abstract class named Animal.

- A concrete method describe(self) that prints a general description of the animal.

```python
import abc 
from abc import ABC ,abstractmethod

class Animal:

    @abstractmethod
    def make_sound(self):
        pass

    def describe(self):
        print(f"General Describition of {self.__class__.__name__} .....")


class Dog(Animal):

    def make_sound(self):
        print("Haw")

class Cat(Animal):

    def make_sound(self):
        print("Naoww")

class Cow(Animal):

    def make_sound(self):
        print("haaaaa")        




d=Dog()
c=Cat()
cc=Cow()


d.make_sound()
c.make_sound()
cc.make_sound()

d.describe()
c.describe()
cc.describe()
                

```

- **Ex:** ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]  → {'Ali': 4, 'Welcome': 1, 'No': 3, 'Hi': 2}

```python
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]

unique=set(words)
result=dict()

for word in unique:
    result[word]=words.count(word)

print(result)    
```

```python
import re
def read_txt_file(file_path):
    ''' Read the contents of a text file an returns it as a string'''
    try:
        with open(file_path,'r',encoding='utf-8')as file:
            content = file.read()
            return content
    except FileNotFoundError:
        return f"error: the file {file_path} was not found"
    except IOError:
        return "error: an aerror occurred while reading the file "

class UserExtractor:
    def __init__(self,file_path):
        self.file_path=file_path
        self.usernames={}

    def extract_usernames(self):
        """Extracts usernames from the text file and stores them in a dictionary."""
        content=read_txt_file(self.file_path)
        usernames= (re.split("[\n:]",content)[::2])
        for i,username in enumerate(usernames):
            self.usernames[i]=username
        return self.usernames
    

ex=UserExtractor("usernames&passwords.txt")
print(ex.extract_usernames())   
```
