````python
from abc import ABC, abstractclassmethod
# abc is a build in python module -> abstract base class
#  ABC is a class we inhert from when we want to create abstract class
# abstractclassmethod is a decorator inforce any child class must implement this method 
# @ is used for decorators  Apply this decorator to the thing below me.
# A decorator is basically a way to add a special behavior or rule to a function or method without changing its code.


class Animal(ABC):
    @abstractclassmethod
    ## inforce child class to implement this function
    def move(self):
        pass

class Bird(Animal):

    def move(self):
        print("move from bird")

class Cat(Animal):
    def move(self):
        print("move from cat")            


a=Bird()
a.move()
````

````python
class Rectangle:
    # one underscore -> mean protected attribute 
    # you shouldn't run it from outside the calls but you can run it outside the class 
    # The _ does not make the variable truly private. 
    # Signal to other programmers
    _length=5
    # dunder varibale make the variable private 
    __width = 10

    def area(self):
        return self._length * self.__width

r=Rectangle()
print(r.area())    
print(r._length)
print(r._Rectangle__width)

#  ============================================ -> name mangling , name corvension
````

````python
class Rectangle:
    def __init__(self,length,width):
        self.__length=length
        self.width=width
    # property decorator let you use a method like an attribute 
    @property
    def length(self):
        return self.__length
    @length.setter
    def length(self,value):
        if value<=0 : raise ValueError("Length must be positive")
        self.__length=value    


r=Rectangle(10,20)
print(r.length)
r.length=77
print(r.length)
````

# `__eq__`, `__str__`, and `__repr__` in OOP

<details>
<summary><strong>__eq__ — Compare Objects</strong></summary>

`__eq__` defines what happens when you use:

```python
obj1 == obj2
```

Example:

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
```

Now:

```python
p1 = Point(1, 2)
p2 = Point(1, 2)

print(p1 == p2)
```

Output:

```text
True
```

**Role:** Defines how two objects are considered equal.

</details>

<details>
<summary><strong>__str__ — User-Friendly Output</strong></summary>

`__str__` defines what appears when you use:

```python
print(obj)
```

Example:

```python
def __str__(self):
    return f"Point({self.x}, {self.y})"
```

Then:

```python
print(p1)
```

Output:

```text
Point(1, 2)
```

**Role:** Provides a **readable representation for users**.

</details>

<details>
<summary><strong>__repr__ — Developer Representation</strong></summary>

`__repr__` defines the object's **official/debug representation**.

Example:

```python
def __repr__(self):
    return f"Point(x={self.x}, y={self.y})"
```

Then:

```python
print(repr(p1))
```

Output:

```text
Point(x=1, y=2)
```

It is mainly useful for **debugging and developers**.

**Role:** Provides a clear, detailed representation of the object.

</details>

<details>
<summary><strong>Quick Summary</strong></summary>

| Method     | Used for       | Main purpose                   |
| ---------- | -------------- | ------------------------------ |
| `__eq__`   | `obj1 == obj2` | Compare objects                |
| `__str__`  | `print(obj)`   | User-friendly output           |
| `__repr__` | `repr(obj)`    | Developer/debug representation |

Easy way to remember:

```text
__eq__   → "Are they equal?"
__str__  → "Show it nicely to the user"
__repr__ → "Show me what this object is"
```

</details>


````python
class Square(Shape):
    def __init__(self,side):
        self.side=side
    def calculate_area(self):
        return self.side**2
    def __str__(self):
        return(f"This the square with side lenght :{self.side}")
    # redefine the role of == operator instead of comparing the two objects address we redifined the operator to compare two objects side

    # need to print(objectname)
    # used to print friendly message
    def __eq__(self, other):
        return self.side==other.side
    # Don't need to use print(objectname )
    # used for developers
    def __repr__(self):
        return(f"Square(side={self.side})")
````

````python
x=circle(5)
s1=Square(5)
s2=Square(5)
print("Area : ",x.calculate_area())
print("Area : ",s1.calculate_area())
print(s1)
print(s1==s2)
s1
````

````python
class Model:

    def fit(self , x, y):
        raise NotImplementedError
    def predict(self,x):
        raise NotImplementedError

class LinearRegression(Model):

    def fit(self , x,y):
        print("Fitting linear regression")

    def predict(self,x):
        print("Predicting linear regression")        


class DicisionTree(Model):
    def fit(self, x, y):
        print("fitting Decision Tree ")
   
    def predict(self, x):
        print("predicting with Decision Tree ")


models =[LinearRegression(),DicisionTree()]

for model in models:
    model.fit(None,None)
    model.predict(None)
````
